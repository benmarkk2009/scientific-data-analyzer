```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print("Data science environment ready!")
```

```python
# Generate sample data
days = np.arange(1, 31)
measurements = 20 + 0.5 * days + np.random.normal(0, 2, 30)

# Create a Data Frame
df = pd.DataFrame({
    "Day": days,
    "Measurement": measurements
})

# Display first five rows
df.head()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Day"], df["Measurement"])
plt.xlabel("Day")
plt.ylabel("Measurement")
plt.title("Sample Data Analysis")
plt.grid(True)

plt.show()
```

```python
from scipy.stats import linregress

#Calculate linear regression
result = linregress(df["Day"], df["Measurement"])

#Display results
print(f"Slope: {result.slope:.4f}")
print(f"Intercept: {result.intercept: .4f}")
print(f"R-squared: {result.rvalue**2: .4f}")
```

```python
# Calculate predicted values
predicted = result.slope * df["Day"] +result.intercept

plt.figure(figsize=(8, 5))

plt.scatter(df["Day"], df["Measurement"], label="Experimental Data")
plt.plot(df["Day"], predicted, color="red", label="Linear Fit")

plt.xlabel("Day")
plt.ylabel("Measurement")
plt.title("Linear Regression Analysis")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#Calculate residuals
df["Predicted"] = predicted
df["Residual"] = df["Measurement"] - df["Predicted"]

#Display results
print(df.head())

#Plot residuals
plt.figure(figsize=(8,5))

plt.scatter(df["Day"], df["Residual"])
plt.axhline(y=0, color="red", linestyle="--")

plt.xlabel("Day")
plt.ylabel("Residual")
plt.title("Residual Analysis")
plt.grid(True)

plt.show()
```
